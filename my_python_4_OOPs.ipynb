### Destructor(destroying object):nothing but deleting objects
 - by using __del__keyword

```python
# Class definition
class Student:
    # Class variable (shared by all instances)
    college = 'XYZ College'

    def __init__(self, roll_no, name, dept, performance):
        # Instance variables
        self.roll_no = roll_no
        self.name = name
        self.dept = dept
        self.performance = performance

    def __del__(self):
        print(f'Destructor called for {self.name}')

    def fullstack(self):
        print(f'{self.name} is a fullstack developer')

    def backend(self):
        print(f'{self.name} is a backend developer')

    def frontend(self):
        print(f'{self.name} is a frontend developer')


# Main program
print(Student.college)  # Accessing class variable

# Creating objects
s1 = Student(1, 'Giri', 'AI', 'Worst')
s2 = Student(2, 'Vishu', 'DS', 'Best')

# Accessing instance attributes
print(s1.name)
print(s2.name)

# Calling methods
Student.fullstack(s1)  # Using class reference to call the method
s2.backend()           # Using object reference to call the method
s1.frontend()

# Explicitly deleting objects to trigger destructor
del s1  # Destructor will be called for s1
del s2  # Destructor will be called for s2

# Program ends here; no further code executes after this point

#print(s1.name)  ##NameError: name 's1' is not defined
```

### Encapsulation
- Public : Accessible from anywhere (default).
- Protected : Indicated by a single underscore (_attribute). Intended for internal use.
- Private : Indicated by double underscores (__attribute). Name-mangled to prevent accidental access.


-The private attribute __pin is encapsulated within the class, ensuring that it cannot be accessed or modified directly from outside the class.
- Controlled access is provided through the display_pin method.

```python
class BankAccount:
    def __init__(self, owner, balance, pin):
        self.owner = owner  # Public
        self._balance = balance  # Protected
        self.__pin = 1234  # Private

    def display_pin(self):
        return self.__pin
        print(BankAccount.display_pin())
```

```python
class BankAccount:
    def __init__(self, owner, balance, pin):
        self.owner = owner  # Public attribute
        self._balance = balance  # Protected attribute
        self.__pin = pin  # Private attribute (uses the pin argument)

    def display_pin(self):
      return self.__pin #  # Access the private attribute



# Create an instance of BankAccount
account = BankAccount("Alice", 1000, 4321)

# Call the instance
print(account.owner)
print(account.display_pin()) # for private class we have to add display__pin(instance)
```

### Abstraction
process of hiding the implementation details of a class and exposing only the essential features or functionalities to the user.
1. Abstract Class :
  - An abstract class is a class that cannot be instantiated directly.
 - It serves as a blueprint for other classes and may contain both abstract methods (methods without implementation) and concrete methods (methods with implementation).
2. Abstract Method :
  - An abstract method is a method declared in an abstract class but does not have any implementation.
  - Subclasses of the abstract class must provide implementations for all abstract methods.
3. Concrete Class :
 - A concrete class is a subclass of an abstract class that implements all the abstract methods defined in the abstract class.
 - Concrete classes can be instantiated.


```python
from abc import ABC, abstractmethod

# abstract class
class Employee(ABC):

    @abstractmethod
    def sector(self):
        pass

class GovtEmployee(Employee):

    # Overriding abstract method
    def sector(self):
        print("I work in Government sector")

class PrivateEmployee(Employee):

    # Overriding abstract method
    def sector(self):
        print("I work in Private sector")

emp1 = GovtEmployee()
emp2 = PrivateEmployee()

emp1.sector()
emp2.sector()
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

print(Circle(10).area())
```

```python
from abc import ABC, abstractmethod

# Abstract Class
class Vehicle(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def start(self):
        """Abstract method: Must be implemented by subclasses"""
        pass

    @abstractmethod
    def stop(self):
        """Abstract method: Must be implemented by subclasses"""
        pass

    def display_info(self):
        """Concrete method: Can be used directly by subclasses"""
        print(f"This is a {self.name}.")

# Concrete Class 1
class Car(Vehicle):
    def start(self):
        print(f"{self.name} is starting.")

    def stop(self):
        print(f"{self.name} is stopping.")

# Concrete Class 2
class Bike(Vehicle):
    def start(self):
        print(f"{self.name} is starting.")

    def stop(self):
        print(f"{self.name} is stopping.")

# Main Program
car = Car("Toyota")
bike = Bike("Honda")

# Using methods
car.display_info()
car.start()
car.stop()

bike.display_info()
bike.start()
bike.stop()
```

### Composition
- Composition models a "has-a" relationship between objects.
- It involves creating objects that contain other objects as attributes to achieve their functionality.

```python
class Engine:
    def __init__(self, type):
        self.type = type  # e.g., "Gasoline", "Electric"

    def start(self):
        print(f"{self.type} engine started.")

    def stop(self):
        print(f"{self.type} engine stopped.")


class Car:
    def __init__(self, brand, engine_type):
        self.brand = brand
        self.engine = Engine(engine_type)  # Composition: Car "has-an" Engine

    def start(self):
        print(f"Starting the {self.brand} car...")
        self.engine.start()

    def stop(self):
        print(f"Stopping the {self.brand} car...")
        self.engine.stop()


# Create a Car with an Electric engine
tesla = Car("Tesla", "Electric")

# Use the Car's methods
tesla.start()
tesla.stop()
```

```python
class Employee:
    retirement_age = 50

    def __init__(self, name, id):
        self.name = name
        self.id = id
        self.salary = None
        self.remaining_years = None

    def years_to_work(self, present_age):
        """Calculate the remaining years to work until retirement."""
        self.remaining_years = self.retirement_age - present_age
        return self.remaining_years

    def setSalary(self, salary):
        """Set the initial salary of the employee."""
        self.salary = salary

    def getSalary(self):
        """Return the current salary of the employee."""
        return self.salary

    def calculate_total_salary(self, present_age):
        """Calculate the total salary earned until retirement with annual increments."""
        if self.salary is None:
            raise ValueError("Salary must be set before calculating total salary.")

        total_salary = 0
        current_salary = self.salary
        remaining_years = self.years_to_work(present_age)

        for _ in range(remaining_years):
            total_salary += current_salary
            current_salary += 10000  # Increment salary by 10,000 every year

        return total_salary


# Create an instance of Employee
emp1 = Employee('Giri', 7)

# Set the initial salary
emp1.setSalary(20000)

# Print the current salary
print("Current Salary:", emp1.getSalary())

# Calculate and print the remaining years to work
present_age = 23
remaining_years = emp1.years_to_work(present_age)
print("Remaining Years to Work:", remaining_years)

# Calculate and print the total salary until retirement
total_salary = emp1.calculate_total_salary(present_age)
print("Total Salary Until Retirement:", total_salary)
```

### Types of Constructor in Python
We have three types of constructors in Python:
1. Non-Parameterized,
2. Parameterized, and

```python
# Non-Parameterized Constructor

class Dress:
   def __init__(self):
       self.cloth = "Cotton"
       self.type = "T-Shirt"
   def get_details(self):
       return f"Cloth = {self.cloth}\nType = {self.type}"
t_shirt = Dress()
print(t_shirt.get_details())
```

```python
# Parameterized Constructor

class Dress:
   def __init__(self, cloth, type, quantity=5):
       self.cloth = cloth
       self.type = type
       self.quantity = quantity
   def get_details(self):
       return f"{self.quantity} {self.cloth} {self.type}(s)"
shirt = Dress("silk", "shirt", 20)
t_shirt = Dress("cotton", "t-shirt")
print(shirt.get_details())
print(t_shirt.get_details())
```

```python
# default constructor

class Dress:
   cloth = "silk"
   type = "shirt"
   def details(self):
       return f"A {self.cloth} {self.type}"
shirt = Dress()
print(shirt.details())
```

```python
# multiple inheritance
class Vodafone:
   subscribers1 = 200
class Idea:
   subscribers2 = 82
class Vi(Vodafone, Idea):
   def subscribers(self):
       print(f"Vi Subscribers = "
             f"{self.subscribers1 + self.subscribers2}"
             f" million")
company = Vi()
company.subscribers()
```

```python
class Fruit:

   def __init__(self, num):
       self.num = num

   def __add__(self, other):   #The __add__ method now handles only two operands (self and other).
       sum = self.num + other.num
       return f"{self.num} fruits + {other.num} fruits = "\
               f"{sum} fruits"

obj1 = Fruit(5)
obj2 = Fruit(4)
obj3 = Fruit(3)
objs = obj1 + obj2
print(objs)
```

```python
class Fruit:
    def __init__(self, num):
        self.num = num  # Number of fruits

    def __add__(self, other):
        # Add the number of fruits from self and other
        total = self.num + other.num
        return Fruit(total)  # Return a new Fruit object with the summed value

#When you write obj1 + obj2 + obj3, Python first evaluates obj1 + obj2,
#which returns a new Fruit object. Then, it adds obj3 to this intermediate result.
    def __str__(self):
        # Define how the object should be represented as a string
        return f"{self.num} fruits"

# Create instances of Fruit
obj1 = Fruit(5)
obj2 = Fruit(4)
obj3 = Fruit(3)
obj4 = Fruit(2)
obj5 = Fruit(1)

# Chain the addition operations
result = obj1 + obj2 + obj3 + obj4 + obj5

# Print the result
print(f"{obj1} + {obj2} + {obj3} {obj4} + {obj5} = {result}")
```

```python
class Lists:

   def __init__(self, li):
       self.li = li

   def __add__(self, other):
       dic = {x:y for x,y in zip(self.li, other.li)}
       return dic

li1 = Lists([1,2,3])
li2 = Lists(["one", "two", "three"])
print(li1 + li2)
```
