```python
try:
  a=int(input("enter the num1 "))
  b=int(input("enter the num2 "))
  c=a/b
  d=a*b
  e=a+b
  print(c)
  print(d)
  print(e)
except NameError:
  print('the user have not defined the variable')
except TypeError:
  print('try to make the datatype similar')
except ZeroDivisionError:
  print("please provide number greater than 0")
except Exception as ex:
  print(ex)
```

```python
try: # try else
  a=int(input("enter the num1 "))
  b=int(input("enter the num2 "))
  c=a/b
  d=a*b
  e=a+b

except NameError:
  print('the user have not defined the variable')
except TypeError:
  print('try to make the datatype similar')
except ZeroDivisionError:
  print("please provide number greater than 0")
except Exception as ex:
  print(ex)
else:
  print(c)
  print(d)
  print(e)
finally:
  print('the excution is done')
```

```python
try: # try else
  a=int(input("enter the num1 "))
  b=int(input("enter the num2 "))
  c=a/b
  d=a*b
  e=a+b

except NameError:
  print('the user have not defined the variable')
except TypeError:
  print('try to make the datatype similar')
except ZeroDivisionError:
  print("please provide number greater than 0")
except Exception as ex:
  print(ex)
else:
  print(c)
  print(d)
  print(e)
finally:
  print('the excution is done')
```

```python
class Error(Exception):
  pass

class dobException(Error):
  pass
```

```python
year=int(input('enter the year of birth '))
age=2024-year
try:
  if age<=30 & age>20:
    print('the age is valid ,you can apply for the exam')
  else:
    raise dobException
except dobException:
  print('the age is not valid,u canot apply for the xam')
```
