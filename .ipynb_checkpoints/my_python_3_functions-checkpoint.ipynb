### **kwargs

```python
def greet(name, age):
    print(f"Hello {name}, you are {age} years old.")

# Calling the function with keyword arguments
greet(age=25, name="Alice")
```

```python
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Alice",age=30, city="New York")
```

### Default Arguments
Default arguments are related to keyword arguments. They allow you to define default values for parameters in a function definition. If a value is not provided for that parameter during the function call, the default value is used.

```python
def greet(name, age, city="Unknown"):
    print(f"Hello {name}, you are {age} years old and live in {city}.")

# Using default value for city
greet("Alice", 25)

# Overriding the default value
greet("Bob", 30, city="Los Angeles")
```

### Arbitrary Argument in Python
Multiple arguments which are passed into a single indefinite length tuple are called arbitrary arguments. We use arbitrary arguments when we don’t know the number of arguments that will be passed to the function.



```python
class Dress:
   def __init__(self, *types):
       self.types = types
   def details(self):
       return self.types
t_shirt = Dress("shirt", "t-shirt")
print(t_shirt.details())
```

```python
def min_max(numbers):
    yield min(numbers), max(numbers), sum(numbers)
    print("This line will not be executed")


numbers = [1, 2, 3, 4, 5]

for smal in min_max(numbers):
  print(smal)
```

### recursion
Recursion involves two main components:

- Base Case : The condition under which the recursion stops. Without a base case, the function would call itself indefinitely, leading to a stack overflow error.
-Recursive Case : The part of the function where it calls itself with modified arguments, moving closer to the base case.
    - Base Case: 0! = 1 (by definition)
    - Recursive Breakdown:
n! = n × (n-1)!


```python
def factorial(n):
    # Base Case
    if n == 0:
        return 1
    # Recursive Case
    else:
        return n * factorial(n - 1)

# Example Usage
print(factorial(3))
```

```python
def fibonacci(n):
    # Base Cases
    if n == 0:
        return 0
    elif n == 1:
        return 1
    # Recursive Case
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

# Example Usage
print(fibonacci(6))
```

### Optimizing Recursion: Memoization
- To improve performance, you can use memoization , which stores the results of expensive function calls and reuses them when the same inputs occur again.
- This version avoids redundant calculations by storing intermediate results in the memo dictionary.

Example: Optimized Fibonacci with Memoization

```python
def fibonacci_memo(n, memo={}):
    # Base Cases
    if n == 0:
        return 0
    elif n == 1:
        return 1
    # Check if result is already computed
    if n in memo:
        return memo[n]
    # Recursive Case with Memoization
    memo[n] = fibonacci_memo(n - 1, memo) + fibonacci_memo(n - 2, memo)
    return memo[n]

# Example Usage
print(fibonacci_memo(60))  # (computed efficiently)
```

### higher-Order Functions

```python
def apply(func, value):
    return func(value)

def square(x):
    return x ** 2

result = apply(square, 4)
print(result)
```

```python
def apply_function(func, value):
    # Apply the given function to the value
    return func(value)

# Define some simple functions
def square(x):
    return x ** 2

def increment(x):
    return x + 1

# Use the higher-order function
print(apply_function(square, 5))
print(apply_function(increment, 5))
```

```python
def create_multiplier(factor):
    # Define a new function inside and return it
    def multiplier(x):
        return x * factor
    return multiplier

# Create specific multiplier functions
double = create_multiplier(2)
triple = create_multiplier(3)

# Use the returned functions
print(double(5))  # Output: 10
print(triple(5))  # Output: 15
```

### Closure
When a function returns another function, the returned function often "remembers" variables from the enclosing scope. This is called a closure .

```python
def outer_function(message):
    def inner_function():
        print(f"Message from outer function: {message}")
    return inner_function

# Create a closure
my_closure = outer_function("Hello, World!")
my_closure()  # Output: Message from outer function: Hello, World!
```

### Custom Higher-Order Functions
You can define your own higher-order functions to encapsulate reusable logic.

Example: A Decorator (Higher-Order Function)
Decorators are a common use case for higher-order functions. A decorator takes a function, modifies its behavior, and returns it.

- Custom higher-order functions, such as decorators, allow you to modify or extend the behavior of existing functions.

```python
def log_function_call(func):
    def wrapper(*args, **kwargs):
        print(f"Calling function {func.__name__} with arguments {args} and {kwargs}")
        result = func(*args, **kwargs)
        print(f"Function {func.__name__} returned {result}")
        return result
    return wrapper

@log_function_call
def add(a, b):
    return a + b

# Use the decorated function
print(add(3, 5))
```

```python
def even_odd(num):
  if num%2==0:
    print('even')
  else:
    print('odd')

#num=int(input('enter a number'))
even_odd(4)
even_odd(5)
```

```python
def is_prime(number):
    # Handle edge cases
    if number <= 1:
        return False  # Numbers less than or equal to 1 are not prime
    if number == 2:
        return True  # 2 is the smallest prime number
    if number % 2 == 0:
        return False  # Even numbers greater than 2 are not prime

    # Check divisors from 3 to the square root of the number
    for i in range(3, int(number**0.5) + 1, 2):
        if number % i == 0:
            return False  # Found a divisor, so the number is not prime

    return True  # No divisors found, so the number is prime


# Input: Get the number from the user
num = int(input("Enter a number: "))

# Process: Check if the number is prime
if is_prime(num):
    print(f"{num} is a prime number.")
else:
    print(f"{num} is not a prime number.")
```

```python
# with no argument:
for i in zip():
    print(i)
```

```python
# 1 argument
for i in zip([1,3,5,7,9]):
  print(i)
```

```python
# two arguments of the same length
for i in zip(('a','b','c','d'),(1,2,3,4)):
  print(i)
```

```python
# two arguments of different length
for i in zip(('a','b','c','d'),(1,2,3)):
  print(i)

for i in zip('Python',[5,4,3,2,1]):
    print(i)
```
