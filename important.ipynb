# Decorators

```python
def my_decor(func):

    def my_wrap(*args, **kwargs):
        print("Decorator Function")
        return func(*args, **kwargs)

    return my_wrap

def my_function(str1, str2):
    print("Main Function")
    print(str1 + " are " +  str2)

my_function = my_decor(my_function)

my_function("Mangoes", "Delicious")
```

```python
def my_decor(func):

    def my_wrap(*args, **kwargs):
        print("Decorator Function")
        return func(*args, **kwargs)

    return my_wrap

@my_decor
def my_function(str1, str2):
    print("Main Function")
    print(str1 + " are " +  str2)

my_function("Mangoes", "Delicious")
```

```python
# Define the decorator
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()  # Call the original function
        print("Something is happening after the function is called.")
    return wrapper

# Apply the decorator using @ syntax
@my_decorator
def say_hello():
    print("Hello!")

# Call the decorated function
say_hello()
```

```python
def my_decor(func):

    def my_wrap(*args, **kwargs):
        print("Decorator Function 1")
        return func(*args, **kwargs)

    return my_wrap

def my_another_decor(func):

    def my_wrap(*args, **kwargs):
        print("Decorator Function 2")
        return func(*args, **kwargs)

    return my_wrap

@my_decor
@my_another_decor
def my_function(str1, str2):
    print("Main Function")
    print(str1 + " are " +  str2)

my_function("Mangoes", "Delicious")
```

### @staticmethod
- Does not take self or cls as its first argument.- 
Cannot access or modify the class state or instance state
### @classmethod
- Takes cls as its first argument.- 
Can access and modify class-level attributes or methods
### @property
- Makes a method callable as an attribute (without parentheses).- 
Often used to implement getters, setters, and deleters for managing attribute access..

```python
class MyClass:
    @staticmethod
    def static_method(arg1, arg2):
        # Perform some operation
        pass
```

```python
class MathOperations:
    @staticmethod
    def add(a, b):
        return a + b

# Call the static method without creating an instance
result = MathOperations.add(5, 3)
print(result)  
```

```python
class MyClass:
    @classmethod
    def class_method(cls, arg1, arg2):
        # Perform some operation
        pass
```

```python
class Person:
    species = "Homo sapiens"

    @classmethod
    def get_species(cls):
        return cls.species

    @classmethod
    def create_person(cls, name):
        return cls(name)

    def __init__(self, name):
        self.name = name

# Access the class method without creating an instance
print(Person.get_species())  

# Use a class method to create an instance
person = Person.create_person("Alice")
print(person.name)  
```

```python
class MyClass:
    @property
    def attribute_name(self):
        # Return the value of the attribute
        pass

    @attribute_name.setter
    def attribute_name(self, value):
        # Set the value of the attribute
        pass

    @attribute_name.deleter
    def attribute_name(self):
        # Delete the attribute
        pass
```

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def radius(self):
        return self._radius

    @radius.setter
    def radius(self, value):
        if value <= 0:
            raise ValueError("Radius must be positive")
        self._radius = value

    @radius.deleter
    def radius(self):
        print("Deleting radius")
        del self._radius

# Create an instance
circle = Circle(5)

# Access the radius property
print(circle.radius)  # Output: 5

# Modify the radius property
circle.radius = 10
print(circle.radius)  # Output: 10

# Attempt to set an invalid radius
try:
    circle.radius = -1
except ValueError as e:
    print(e)  # Output: Radius must be positive

# Delete the radius property
del circle.radius  # Output: Deleting radius
```

## Generators
- when working with large datasets or infinite sequences, as they produce values one at a time and only when needed.

- Generators are implemented using either:
  - Generator functions (functions that use the yield keyword).
  - Generator expressions (similar to list comprehensions but with parentheses instead of       square brackets).

- Key Characteristics of Generators
  - Lazy Evaluation :
     - Values are generated one at a time and only when requested.
     - This saves memory compared to creating an entire list of values upfront.
  - Stateful :
     - A generator remembers its state between calls, allowing it to resume execution where it left off.
  - Iterables :
     - Generators are iterators, meaning they implement the __iter__() and __next__() methods automatically.
  - One-Time Use :
     - Once a generator is exhausted (all values have been generated), it cannot be reused. You need to recreate it if you want to iterate over it again.

1. Using a Generator Function
- A generator function is defined like a regular function, but instead of return, it uses the yield keyword to produce values.

```python
def my_generator():
    yield 1
    yield 2
    yield 3,2,1

# Create a generator object
gen = my_generator()

# Iterate over the generator
for value in gen:
    print(value)
```

```python
def my_generator():
    return 1,2,3
    return 'pyth'
   

# Create a generator object
gen = my_generator()

# Iterate over the generator
for value in gen:
    print(value)
```

```python
# List comprehension (creates a list in memory)
squares_list = [x**2 for x in range(5)]
print(squares_list)  # Output: [0, 1, 4, 9, 16]

# Generator expression (creates a generator)
squares_gen = (x**2 for x in range(5))

# Iterate over the generator
for square in squares_gen:
    print(square)
```

```python
# List comprehension (creates a list in memory)
squares_list = [x**2 for x in range(5)]
print(squares_list)  # Output: [0, 1, 4, 9, 16]

# Generator expression (creates a generator)
squares_gen = (x**2 for x in range(5))

val=list(squares_gen)
print(val)


# Iterate over the generator
for square in val:
    print(square)
```

## Iterators in Python
- iterators implemented using class where as generators using fonction
- Python objects that iterate through iterable objects are called Iterators. It is used to iterate over objects by returning one value at a time. Iterators are created by using the iter() function. The function next() is used to get the subsequent value from the iterator.

```python
nums = [1, 2, 3, 4]
obj = iter(nums)
print(next(obj))
print(next(obj))
print(next(obj))
print(next(obj))
```

```python
class Integers:

  def __init__(self, max_limit):
      self.max_limit = max_limit

  def __iter__(self):
      self.num = 1
      return self

  def __next__(self):
      temp = self.num
      self.num += 1
      return temp

my_integers = Integers(0)
my_iterator = iter(my_integers)
print(next(my_iterator))
print(next(my_iterator))
print(next(my_iterator))
print(next(my_iterator))
print(next(my_iterator))
print(next(my_iterator))
```

```python
class Multiples:

  def __iter__(self):
      self.val = 1
      return self

  def __next__(self):
      temp = self.val
      self.val += 1
      return temp*5

multiples5 = Multiples()
obj = iter(multiples5)

print(next(obj))
print(next(obj))
print(next(obj))
```
